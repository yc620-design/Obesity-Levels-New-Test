```python
print("Duplicate rows before:", obe.duplicated().sum())

obe = obe.drop_duplicates()

print("Duplicate rows after:", obe.duplicated().sum())
```

```python
# Calculating Outliers for the 'Weight' column using the IQR Method
Q1 = obe_cleaned['Weight'].quantile(0.25)
Q3 = obe_cleaned['Weight'].quantile(0.75)
IQR = Q3 - Q1

# Define the lower and upper bounds for outliers
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Filter the dataset to find the outlier rows
outliers_weight = obe_cleaned[(obe_cleaned['Weight'] < lower_bound) | (obe_cleaned['Weight'] > upper_bound)]

print("--- Outlier Detection using IQR ---")
print(f"Lower Bound for Weight: {lower_bound:.2f}")
print(f"Upper Bound for Weight: {upper_bound:.2f}")
print(f"Total number of statistical outliers in Weight: {outliers_weight.shape[0]} rows")

# Note for report:
# Since this is an Obesity dataset, "extreme" weights are actual target classes (e.g., Obesity Type III).
# Therefore, dropping these outliers would result in data loss for critical categories. We will keep them.
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# 1. Encode the target variable: Convert 'NObeyesdad' (Obesity Level) to numeric labels (0, 1, 2, 3...)
le = LabelEncoder()
obe_cleaned['Target'] = le.fit_transform(obe_cleaned['NObeyesdad'])

# Print the mapping dictionary for your report reference
obesity_mapping = dict(zip(le.classes_, le.transform(le.classes_)))
print("Numeric mapping for Obesity Levels:\n", obesity_mapping)

# 2. Separate features (X) and the target variable (y)
X = obe_cleaned.drop(columns=['NObeyesdad', 'Target'])
y = obe_cleaned['Target']

# 3. Apply One-Hot Encoding to categorical features (e.g., Gender, SMOKE)
# drop_first=True prevents the dummy variable trap (multicollinearity)
X_encoded = pd.get_dummies(X, drop_first=True)

print("\nFirst two rows of the encoded features (X_encoded):")
display(X_encoded.head(2))
```

```python
from sklearn.preprocessing import StandardScaler

# Initialize the StandardScaler
scaler = StandardScaler()

# Apply standardization to scale all features to a similar range
# (Mean = 0, Standard Deviation = 1)
X_scaled = pd.DataFrame(scaler.fit_transform(X_encoded), columns=X_encoded.columns)

print("Standardization complete! First two rows of scaled features (X_scaled):")
display(X_scaled.head(2))
```

```python
# ============================================================
# CLEAN OUTLIERS USING IQR CAPPING
# Age, Height, Weight
# ============================================================

import pandas as pd

features = ['Age', 'Height', 'Weight']

for column in features:

    # Calculate Q1 and Q3
    Q1 = obe_cleaned[column].quantile(0.25)
    Q3 = obe_cleaned[column].quantile(0.75)

    # Calculate IQR
    IQR = Q3 - Q1

    # Define lower and upper bounds
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Count outliers before cleaning
    outlier_count = (
        (obe_cleaned[column] < lower_bound) |
        (obe_cleaned[column] > upper_bound)
    ).sum()

    print(f"\n{column}")
    print(f"Lower Bound: {lower_bound:.2f}")
    print(f"Upper Bound: {upper_bound:.2f}")
    print(f"Outliers Before Cleaning: {outlier_count}")

    # --------------------------------------------------------
    # CAPPING METHOD
    # Values below lower bound become lower bound
    # Values above upper bound become upper bound
    # --------------------------------------------------------

    obe_cleaned[column] = obe_cleaned[column].clip(
        lower=lower_bound,
        upper=upper_bound
    )


print("\nOutlier capping completed.")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


from sklearn.model_selection import (
    train_test_split,
    GridSearchCV,
    StratifiedKFold
)

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import ConfusionMatrixDisplay
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)
```

```python
# ============================================================
# TRAIN / TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set:", X_train.shape)
print("Testing set :", X_test.shape)
```

```python
# ============================================================
# STANDARDIZATION
# ============================================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train
)

X_test_scaled = scaler.transform(
    X_test
)
```

```python
# ============================================================
# CROSS VALIDATION
# ============================================================

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
# ============================================================
# MODEL 1 - DECISION TREE
# ============================================================

dt_model = DecisionTreeClassifier(
    random_state=42
)

dt_param_grid = {
    'criterion': [
        'gini',
        'entropy'
    ],

    'max_depth': [
        3,
        5,
        7,
        10
    ],

    'min_samples_split': [
        5,
        10,
        20
    ],

    'min_samples_leaf': [
        3,
        5,
        10
    ]
}

dt_grid = GridSearchCV(
    estimator=dt_model,
    param_grid=dt_param_grid,
    scoring='f1_macro',
    cv=cv,
    n_jobs=-1
)

dt_grid.fit(
    X_train_scaled,
    y_train
)

best_dt = dt_grid.best_estimator_

dt_pred = best_dt.predict(
    X_test_scaled
)
```

```python
# ============================================================
# MODEL 2 - Logistic Regression (Baseline)
# ============================================================

lr_model = LogisticRegression(
    C=1.0,
    solver='lbfgs',
    max_iter=3000,
    penalty='l2',
    random_state=42
)

lr_model.fit(
    X_train_scaled,
    y_train
)

lr_pred = lr_model.predict(
    X_test_scaled
)

best_lr = lr_model
```

```python
# ============================================================
# MODEL 3 - K-NEAREST NEIGHBORS
# ============================================================

knn_model = KNeighborsClassifier()


knn_param_grid = {
    'n_neighbors': [3, 5, 7, 9, 11, 15, 21, 25],
    'weights': ['uniform', 'distance'],
    'metric': ['euclidean', 'manhattan', 'minkowski'],
    'p': [1, 2]
}


knn_grid = GridSearchCV(
    estimator=knn_model,
    param_grid=knn_param_grid,
    scoring='accuracy',
    cv=cv,
    n_jobs=-1
)


knn_grid.fit(
    X_train_scaled,
    y_train
)


# BEST KNN model
best_knn = knn_grid.best_estimator_


# Prediction using BEST parameters
knn_pred = best_knn.predict(
    X_test_scaled
)
```

```python
# ============================================================
# DISPLAY BEST PARAMETERS
# ============================================================

print("\n")
print("=" * 75)
print("BEST PARAMETERS FROM GRIDSEARCHCV")
print("=" * 75)


print("\nDecision Tree:")
print(dt_grid.best_params_)

print(
    "CV Macro F1:",
    f"{dt_grid.best_score_ * 100:.2f}%"
)


print("\nLogistic Regression (Baseline):")

print({
    "C": lr_model.C,
    "solver": lr_model.solver,
    "penalty": lr_model.penalty,
    "max_iter": lr_model.max_iter
})

print(
    "Tuning:",
    "Fixed parameters / No GridSearchCV"
)


print("\nK-Nearest Neighbors:")
print(knn_grid.best_params_)

print(
    "CV Macro F1:",
    f"{knn_grid.best_score_ * 100:.2f}%"
)
```

```python
# ============================================================
# 1. EVALUATION FUNCTION
# ============================================================

def evaluate_model(
    model_name,
    y_true,
    y_pred
):

    accuracy = accuracy_score(
        y_true,
        y_pred
    )

    macro_precision = precision_score(
        y_true,
        y_pred,
        average='macro',
        zero_division=0
    )

    macro_recall = recall_score(
        y_true,
        y_pred,
        average='macro',
        zero_division=0
    )

    macro_f1 = f1_score(
        y_true,
        y_pred,
        average='macro',
        zero_division=0
    )


    return {

        'Model':
            model_name,

        'Accuracy':
            accuracy,

        'Macro Precision':
            macro_precision,

        'Macro Recall':
            macro_recall,

        'Macro F1-Score':
            macro_f1
    }
```

```python
# ============================================================
# EVALUATE BEST MODELS
# ============================================================

dt_result = evaluate_model(
    'Decision Tree',
    y_test,
    dt_pred
)


lr_result = evaluate_model(
    'Logistic Regression',
    y_test,
    lr_pred
)


knn_result = evaluate_model(
    'K-Nearest Neighbors',
    y_test,
    knn_pred
)


results_df = pd.DataFrame([
    dt_result,
    lr_result,
    knn_result
])


display(results_df)
```

```python
# ============================================================
# OVERALL PERFORMANCE COMPARISON GRAPH
# ============================================================

plot_metrics = [
    'Accuracy',
    'Macro Precision',
    'Macro Recall',
    'Macro F1-Score'
]


plot_df = results_df[
    ['Model'] + plot_metrics
].copy()


# Convert to percentage
for col in plot_metrics:

    plot_df[col] = (
        plot_df[col] * 100
    )


plot_df = plot_df.set_index(
    'Model'
)


ax = plot_df.plot(
    kind='bar',
    figsize=(13, 7),
    width=0.75
)


plt.title(
    'Overall Performance Comparison of Machine Learning Models',
    fontsize=15,
    pad=15
)


plt.xlabel(
    'Machine Learning Model',
    fontsize=12
)


plt.ylabel(
    'Performance Score (%)',
    fontsize=12
)


plt.ylim(
    0,
    105
)


plt.xticks(
    rotation=0,
    fontsize=10
)


plt.legend(
    title='Evaluation Metrics',
    loc='lower right'
)


# Add percentage labels
for container in ax.containers:

    labels = []

    for bar in container:

        height = bar.get_height()

        labels.append(
            f'{height:.2f}%'
        )

    ax.bar_label(
        container,
        labels=labels,
        padding=3,
        fontsize=9
    )


plt.tight_layout()

plt.show()
```
